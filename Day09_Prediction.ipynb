```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

print("========== Day 9 - Student Score Prediction ==========")

df = pd.read_csv("clean_student_scores.csv")

print("\nStudent Score Dataset")
display(df)

X = df[["Study_Hours"]]
y = df["Score"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)

predictions = model.predict(X_test)

print("\nActual vs Predicted Scores")

result = pd.DataFrame({
    "Study Hours": X_test["Study_Hours"].values,
    "Actual Score": y_test.values,
    "Predicted Score": predictions.round(2)
})

display(result)

study_hours = pd.DataFrame({"Study_Hours": [7]})

predicted_score = model.predict(study_hours)

print("\nPrediction Example")
print("Study Hours:", study_hours["Study_Hours"][0])
print("Predicted Score:", round(predicted_score[0], 2))

print("\nDay 9 Task Completed Successfully")
```
