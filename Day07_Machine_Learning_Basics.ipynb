```python
import pandas as pd
from sklearn.model_selection import train_test_split

print("========== Day 7 - Machine Learning Basics ==========")

df = pd.read_csv("clean_student_scores.csv")

print("\nStudent Score Dataset")
display(df)

print("\nSupervised Learning")
print("Supervised Learning is a type of Machine Learning where the model learns from labeled data.")
print("In this dataset:")
print("Input (Feature): Study_Hours")
print("Output (Target): Score")

X = df[["Study_Hours"]]
y = df["Score"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

print("\nTrain-Test Split Completed")

print("\nTraining Features")
display(X_train)

print("\nTesting Features")
display(X_test)

print("\nTraining Target")
display(y_train)

print("\nTesting Target")
display(y_test)

print("\nLinear Regression")
print("Linear Regression is a supervised Machine Learning algorithm used to predict continuous values.")
print("It finds the best-fit line between the input feature and the target value.")

print("\nDay 7 Task Completed Successfully")
```
